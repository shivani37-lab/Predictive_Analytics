```python
# ============================================================
# PROJECT 3 - PREDICTIVE ANALYTICS USING HISTORICAL DATA
# Sales Forecasting using Machine Learning
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import urllib.request
import warnings

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")

print("=" * 60)
print("PROJECT 3 - PREDICTIVE ANALYTICS")
print("=" * 60)


# -----------------------------
# 2. DOWNLOAD DATASET IF NEEDED
# -----------------------------

file_path = "Data/Sample - Superstore.csv"

# Download the dataset if the file is missing or empty
if not os.path.exists(file_path) or os.path.getsize(file_path) == 0:

    print("\nDataset missing or empty.")
    print("Downloading dataset...")

    url = "https://raw.githubusercontent.com/eyowhite/Data-Analysis/main/Sample%20-%20Superstore.csv"

    urllib.request.urlretrieve(url, file_path)

    print("Dataset downloaded successfully! ✅")


# -----------------------------
# 3. LOAD DATASET
# -----------------------------

df = pd.read_csv(file_path, encoding="latin1")

print("\nDataset loaded successfully! ✅")

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())


# -----------------------------
# 4. BASIC DATA INFORMATION
# -----------------------------

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())


# -----------------------------
# 5. DATA CLEANING
# -----------------------------

print("\n" + "=" * 60)
print("DATA CLEANING")
print("=" * 60)

# Remove duplicate rows
duplicate_count = df.duplicated().sum()

print("\nDuplicate rows:", duplicate_count)

df = df.drop_duplicates()

# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

# Convert Sales to numeric
df["Sales"] = pd.to_numeric(df["Sales"], errors="coerce")

# Remove rows with missing date or sales
df = df.dropna(subset=["Order Date", "Sales"])

# Sort by date
df = df.sort_values("Order Date")

print("\nData cleaned successfully! ✅")

print("Rows after cleaning:", len(df))


# -----------------------------
# 6. DESCRIPTIVE STATISTICS
# -----------------------------

print("\n" + "=" * 60)
print("SALES STATISTICS")
print("=" * 60)

print(df["Sales"].describe())


# -----------------------------
# 7. MONTHLY SALES AGGREGATION
# -----------------------------

monthly_sales = (
    df.set_index("Order Date")
      .resample("MS")["Sales"]
      .sum()
      .reset_index()
)

monthly_sales.columns = ["Date", "Sales"]

print("\nMonthly Sales Data:")
display(monthly_sales.head(12))


# -----------------------------
# 8. VISUALIZE HISTORICAL SALES
# -----------------------------

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["Date"],
    monthly_sales["Sales"],
    marker="o",
    linewidth=2
)

plt.title("Historical Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# -----------------------------
# 9. CREATE TIME-SERIES FEATURES
# -----------------------------

data = monthly_sales.copy()

# Calendar features
data["Year"] = data["Date"].dt.year
data["Month"] = data["Date"].dt.month
data["Quarter"] = data["Date"].dt.quarter

# Lag features
data["Lag_1"] = data["Sales"].shift(1)
data["Lag_2"] = data["Sales"].shift(2)
data["Lag_3"] = data["Sales"].shift(3)
data["Lag_6"] = data["Sales"].shift(6)
data["Lag_12"] = data["Sales"].shift(12)

# Rolling average
data["Rolling_3"] = data["Sales"].shift(1).rolling(3).mean()
data["Rolling_6"] = data["Sales"].shift(1).rolling(6).mean()

# Remove rows created by lagging
data = data.dropna().reset_index(drop=True)

print("\nFeatures created successfully! ✅")

display(data.head())


# -----------------------------
# 10. PREPARE TRAINING DATA
# -----------------------------

features = [
    "Year",
    "Month",
    "Quarter",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Lag_6",
    "Lag_12",
    "Rolling_3",
    "Rolling_6"
]

X = data[features]
y = data["Sales"]


# -----------------------------
# 11. TIME-BASED TRAIN/TEST SPLIT
# -----------------------------

# Use the first 80% for training
# Use the last 20% for testing

split_index = int(len(data) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# -----------------------------
# 12. TRAIN MACHINE LEARNING MODEL
# -----------------------------

model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    max_depth=10,
    min_samples_split=2
)

model.fit(X_train, y_train)

print("\nRandom Forest model trained successfully! 🤖")


# -----------------------------
# 13. PREDICT TEST DATA
# -----------------------------

y_pred = model.predict(X_test)


# -----------------------------
# 14. MODEL EVALUATION
# -----------------------------

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("\n" + "=" * 60)
print("MODEL EVALUATION")
print("=" * 60)

print(f"\nMean Absolute Error (MAE): {mae:,.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")
print(f"R² Score: {r2:.4f}")


# -----------------------------
# 15. ACTUAL VS PREDICTED SALES
# -----------------------------

results = pd.DataFrame({
    "Date": data.iloc[split_index:]["Date"].values,
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

print("\nActual vs Predicted Sales:")
display(results)


# -----------------------------
# 16. VISUALIZE ACTUAL VS PREDICTED
# -----------------------------

plt.figure(figsize=(14, 6))

plt.plot(
    results["Date"],
    results["Actual Sales"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    results["Date"],
    results["Predicted Sales"],
    marker="o",
    linestyle="--",
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# -----------------------------
# 17. FUTURE SALES FORECAST
# -----------------------------

print("\n" + "=" * 60)
print("FUTURE SALES FORECAST")
print("=" * 60)

# Start with the complete historical monthly dataset
forecast_data = monthly_sales.copy()

future_predictions = []

# Predict next 6 months
for i in range(6):

    next_date = forecast_data["Date"].max() + pd.DateOffset(months=1)

    # Get previous sales values
    sales_values = forecast_data["Sales"].tolist()

    lag_1 = sales_values[-1]
    lag_2 = sales_values[-2]
    lag_3 = sales_values[-3]
    lag_6 = sales_values[-6]
    lag_12 = sales_values[-12]

    rolling_3 = np.mean(sales_values[-3:])
    rolling_6 = np.mean(sales_values[-6:])

    future_features = pd.DataFrame({
        "Year": [next_date.year],
        "Month": [next_date.month],
        "Quarter": [next_date.quarter],
        "Lag_1": [lag_1],
        "Lag_2": [lag_2],
        "Lag_3": [lag_3],
        "Lag_6": [lag_6],
        "Lag_12": [lag_12],
        "Rolling_3": [rolling_3],
        "Rolling_6": [rolling_6]
    })

    prediction = model.predict(future_features)[0]

    # Prevent negative sales prediction
    prediction = max(0, prediction)

    future_predictions.append({
        "Date": next_date,
        "Forecasted Sales": prediction
    })

    # Add prediction for next recursive prediction
    forecast_data = pd.concat([
        forecast_data,
        pd.DataFrame({
            "Date": [next_date],
            "Sales": [prediction]
        })
    ], ignore_index=True)


forecast_df = pd.DataFrame(future_predictions)

print("\nNext 6 Months Forecast:")
display(forecast_df)


# -----------------------------
# 18. VISUALIZE FUTURE FORECAST
# -----------------------------

plt.figure(figsize=(14, 6))

# Historical
plt.plot(
    monthly_sales["Date"],
    monthly_sales["Sales"],
    label="Historical Sales",
    linewidth=2
)

# Forecast
plt.plot(
    forecast_df["Date"],
    forecast_df["Forecasted Sales"],
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Forecasted Sales"
)

plt.title("Historical Sales and Future Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# -----------------------------
# 19. FEATURE IMPORTANCE
# -----------------------------

importance = pd.DataFrame({
    "Feature": features,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    "Importance",
    ascending=False
)

print("\nFeature Importance:")
display(importance)


# -----------------------------
# 20. FEATURE IMPORTANCE CHART
# -----------------------------

plt.figure(figsize=(10, 6))

plt.barh(
    importance["Feature"],
    importance["Importance"]
)

plt.gca().invert_yaxis()

plt.title("Feature Importance in Sales Prediction")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()


# -----------------------------
# 21. FINAL SUMMARY
# -----------------------------

print("\n" + "=" * 60)
print("PROJECT SUMMARY")
print("=" * 60)

print("\n✅ Dataset loaded")
print("✅ Data cleaned and preprocessed")
print("✅ Historical sales analyzed")
print("✅ Monthly sales trends visualized")
print("✅ Time-series features created")
print("✅ Random Forest regression model trained")
print("✅ Model evaluated using MAE, RMSE and R²")
print("✅ Actual vs predicted sales visualized")
print("✅ Future sales forecast generated")
print("✅ Feature importance analyzed")

print("\n🎉 PREDICTIVE ANALYTICS PROJECT COMPLETED!")
```
